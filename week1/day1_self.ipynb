```python
import os
from openai import OpenAI 
from scraper import fetch_website_contents
from dotenv import load_dotenv
from IPython.display import Markdown, display
```

```python
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')
if not api_key:
    print('No API key found!')
elif not api_key.startswith('sk-proj-'):
    print('API key found but does not start with sk-proj-, please check for the correct API key!')
elif api_key.strip() != api_key:
    print('There might be leading or trailing spaces in the key, please check!')
else:
    print('API key found!')
```

```python
message = 'Hello, you name shall be Linus from now onwards, and my name is Rikesh'
messages = [
    {'role': 'system', 'content': 'You are the top professional in the field of AI Engineering. After every response, add a quick information about AI engineering that can help a beginner to become a better AI Engineer and it should be structured.'},
    {'role': 'user', 'content': message}
    ]
messages
```

```python
openai = OpenAI()
response = openai.chat.completions.create(model='gpt-5-nano', messages=messages)
print(response.choices[0].message.content)
```

```python
rm = fetch_website_contents('https://ornate-torte-03c391.netlify.app/')
print(rm)
```

```python
system_prompt = 'You are an expert software engineer. With every reponse you provide, give one structured tip or knowledge or information on tools and techniques and also on how to be a better software engingeer, it can be regarding backend development, frontend, system design or dsa.'
user_prompt = 'Take a look at this website, and provide constructive criticism to the creator.'
print(system_prompt)
print(user_prompt)
```

```python
def messages_for(website):
    return [
    {'role': 'system', 'content': system_prompt},
    {'role': 'user', 'content': user_prompt + website}
    ]
print(messages_for(rm))
```

```python
response = openai.chat.completions.create(model='gpt-4.1-mini', messages=messages_for(rm))
display(Markdown(response.choices[0].message.content))
```
